##Scraping Banyak Halaman - Kopi Kenangan

###Algoritma

1. Tentukan alamat url yang akan discrape (di sini pakai daftar nama halaman, bukan nomor, karena Kopi Kenangan gak punya sistem /page/1, /page/2)
2. Buat array kosong sesuai jumlah feature yang akan dimasukkan (di sini ada 3: nama halaman, judul halaman, ringkasan heading)
3. Buat loop dari setiap halaman yang akan diambil
4. Append terhadap array dari setiap tag yang ditemukan

```python
# LANGKAH 1: tentukan alamat url yang akan discrape
root = 'https://www.kopikenangan.com'
halaman_list = ['/about', '/news', '/career', '/outlets', '/kenangan-academy']
```

```python
# LANGKAH 2: buat array kosong sesuai jumlah feature yang mau diambil
nama_halaman = []
judul_halaman = []
isi_teks = []
```

```python
# LANGKAH 3: buat loop dari setiap halaman
# LANGKAH 4: append terhadap array dari setiap tag yang ditemukan

for halaman in halaman_list:
  url = root + halaman
  response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'})
  soup = BeautifulSoup(response.text, 'html.parser')

  # ambil judul halaman (tag <title>)
  judul = soup.title.text.strip() if soup.title else 'Tidak ada judul'

  # ambil semua heading h1-h3 di halaman itu, gabung jadi satu teks
  headings = soup.find_all(['h1', 'h2', 'h3'])
  teks_gabungan = ' | '.join([h.get_text(strip=True) for h in headings if h.get_text(strip=True)])

  # masukin ke array (append)
  nama_halaman.append(halaman)
  judul_halaman.append(judul)
  isi_teks.append(teks_gabungan)

  print(f"Selesai scrape: {halaman}")

len(nama_halaman), len(judul_halaman), len(isi_teks)

df_multipage = pd.DataFrame({
    'Halaman': nama_halaman,
    'Judul': judul_halaman,
    'Ringkasan_Heading': isi_teks
})
df_multipage
```

**WEB SCRAPING WITH API**

**How to use Wikipedia API?**
1. Go to the Wikipedia API endpoint: (https://en.wikipedia.org/w/api.php?utm_)
2. Wikipedia menyediakan API untuk mengambil berbagai data dari artikel Wikipedia.
3. Untuk request sederhana, tidak perlu membuat atau mendapatkan API key.
4. Gunakan parameter API untuk menentukan data yang ingin diambil. contohnya:
*   action=query → mengambil data
*   format=json → menentuan format hasil sebagai JSON
*   prop=extracts → mengambil isi artikel
*   explaintext=True → mengambil teks biasa tanpa format HTML
*   titles=Data science → menentukan artikel yang ingin diambil
5. Kirim request menggunakan Python dengan library **requests**.
6. Tambahkan User-Agent untuk mengidentifikasi aplikasi atau pengguna yang melakukan request.
7. Hasil response dapat diubah menjadi data Python menggunakan **response.json()**

```python
#mencoba request API
import requests
import json

url = 'https://en.wikipedia.org/w/api.php'
params = {"action": "query", "format": "json", "prop": "extracts", "explaintext": True, "titles": "Data science"}
headers = {"User-Agent": "DataScienceStudent/1.0 (your-email@example.com)"}
response = requests.get(url, params=params, headers=headers)
print(response.status_code)
```

```python
api_request_json = response.json()
api_request_json
```

```python
with open("response.json", "w") as f:
   json.dump(api_request_json, f, indent=4)
```

```python
with open("response.json", "r") as f:
    api_request_json = json.load(f)

query = api_request_json["query"]
pages = query["pages"]
page = list(pages.values())[0]
```

```python
#simpan ke dataframe
import pandas as pd

title = page["title"]
extract = page["extract"]

df = pd.DataFrame({'Title': [title], 'Content': [extract]})
df
```
